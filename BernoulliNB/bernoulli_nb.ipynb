```python
x = data.drop(columns=['class'])
y = data['class']
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42, stratify=y)
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.naive_bayes import BernoulliNB
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
encoder = OneHotEncoder(handle_unknown='ignore')

x_train_enc = encoder.fit_transform(x_train)
x_test_enc = encoder.transform(x_test)

print(x_train_enc.shape, x_test_enc.shape, '\n', x_train_enc[:5].toarray())
```

```python
model = BernoulliNB(alpha=1.0)
model.fit(x_train_enc, y_train)
```

```python
y_pred = model.predict(x_test_enc)

print('accuracy:', accuracy_score(y_test, y_pred))
print('\nconfusion matrix:\n', confusion_matrix(y_test, y_pred))
print('\nclassification report:\n', classification_report(y_test, y_pred))
```

```python
y_proba = model.predict_proba(x_test_enc)
print(model.classes_)
print(y_proba[:5])
```

```python
p_poisonous = y_proba[:, 1]

print('min, max, mean\n', p_poisonous.min(), p_poisonous.max(), p_poisonous.mean())

plt.figure(figsize=(8, 5))
sns.histplot(p_poisonous, bins=30, kde=True)
plt.xlabel('p')
plt.ylabel('number of observations')
plt.show()
```

```python
from sklearn.calibration import calibration_curve

prob_true, prob_pred = calibration_curve(y_test == 'p', p_poisonous, n_bins=10)

plt.figure(figsize=(7, 6))
plt.plot(prob_pred, prob_true, marker="o", label="BernoulliNB")
plt.plot([0, 1], [0, 1], linestyle="--", label="Perfect calibration")
plt.xlabel("Mean predicted probability")
plt.ylabel("Fraction of poisonous observations")
plt.title("Calibration Curve")
plt.legend()
plt.show()
```
